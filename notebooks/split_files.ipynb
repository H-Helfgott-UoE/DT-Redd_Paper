```python
import os
import random
import json
import numpy as np
```

```python
# ================== INPUT/OUTPUT FILES ===================

# Folder containing all raw audio splits
SPLITS_FOLDER = r''

# Output path to splits breakdown
ALL_SPLITS = r''

# ====================== PARAMETERS =======================

# Fractions for calibration and validation sets
TRAIN_FRAC, TEST_FRAC = 0.8, 0.2

# Number of folds in training set
NB_FOLDS = 5
```

### Compute a random 80/20 split, and divide the 80% set into 5 equal-sized folds

```python
file_list = []

for split_folder in sorted(os.listdir(SPLITS_FOLDER)):

    if not os.path.isdir(os.path.join(SPLITS_FOLDER, split_folder)):
        continue

    for file_name in sorted(os.listdir(os.path.join(SPLITS_FOLDER, split_folder))):
        file_list.append([split_folder, file_name])

random.seed(100)
random.shuffle(file_list)

train_length = int(TRAIN_FRAC * len(file_list))
test_length = len(file_list) - train_length

split_dict = {f'train_fold_{k}' : [] for k in range(NB_FOLDS)}

for i, item in enumerate(file_list[:train_length]):
    split_dict[f'train_fold_{i % NB_FOLDS}'].append(item)

split_dict['train'] = file_list[:train_length]
split_dict['test'] = file_list[train_length:]

for key, sub_list in split_dict.items():
    print(key,'— nb files: ',len(sub_list),'('+str(np.round(100 * len(sub_list)/len(file_list), 2))+" %)")
```

### Save the resulting splits breakdown

```python
split_dict['base_folder'] = SPLITS_FOLDER
with open(ALL_SPLITS, 'w') as output:
    json.dump(split_dict, output)
```
